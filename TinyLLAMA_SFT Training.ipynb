```python
# 2. Imports
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, TrainingArguments, Trainer, DataCollatorForLanguageModeling
from datasets import load_dataset
from peft import prepare_model_for_kbit_training, LoraConfig, get_peft_model
```

```python
hf_token = " " #Enter your Hugging Face token here

# 3. Load Dataset (Dolly)
dataset = load_dataset("databricks/databricks-dolly-15k")

# 4. Prepare Prompts
def format_prompt(example):
    return {
        "text": f"### Instruction:\n{example['instruction']}\n\n### Response:\n{example['response']}"
    }

formatted_dataset = dataset.map(format_prompt)
train_data = formatted_dataset["train"].train_test_split(test_size=0.1, seed=42)
```

```python
# 5. Load Base Model and Tokenizer
base_model = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(base_model)
tokenizer.pad_token = tokenizer.eos_token  # Required for DataCollator

model = AutoModelForCausalLM.from_pretrained(
    base_model,
    torch_dtype=torch.bfloat16,
    device_map="auto"
)
```

```python
# 6. Apply LoRA
#Trial 1
lora_config = LoraConfig(
    r=8,
    lora_alpha=32,
    target_modules=["q_proj", "v_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM"
)

'''
TRIAL_CONFIGS = [
    # Trial 1
    {
        "lora_rank": 8,
        "lora_alpha": 32,
        "lora_dropout": 0.05,
        "target_modules": ["q_proj", "v_proj"],
        "lr": 2e-4,
        "batch_size": 8,
        "epochs": 2
    },
    # Trial 2
    {
        "lora_rank": 16,
        "lora_alpha": 64,
        "lora_dropout": 0.1,
        "target_modules": ["q_proj", "v_proj", "k_proj", "o_proj"],
        "lr": 1e-4,
        "batch_size": 8,
        "epochs": 2
    },
    # Trial 3
    {
        "lora_rank": 32,
        "lora_alpha": 128,
        "lora_dropout": 0.2,
        "target_modules": ["q_proj", "v_proj"],
        "lr": 5e-5,
        "batch_size": 4,
        "epochs": 4
    },
    # Trial 4
    {
        "lora_rank": 64,
        "lora_alpha": 256,
        "lora_dropout": 0.1,
        "target_modules": ["q_proj", "v_proj", "k_proj", "o_proj", "gate_proj"],
        "lr": 3e-4,
        "batch_size": 8,
        "epochs": 3
    },
    # Trial 5
    {
        "lora_rank": 8,
        "lora_alpha": 64,
        "lora_dropout": 0.05,
        "target_modules": ["gate_proj", "up_proj", "down_proj"],
        "lr": 2e-5,
        "batch_size": 4,
        "epochs": 5
    }
]

'''


model = get_peft_model(model, lora_config)

# 7. Tokenize Data
def tokenize(example):
    return tokenizer(example["text"], truncation=True, padding="max_length", max_length=256)

tokenized_data = train_data.map(tokenize, batched=True)

# 8. Data Collator
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

```

```python
# 9. Training Arguments (Trial 1 Example)
training_args = TrainingArguments(
    output_dir="./tinyllama-lora-sft",
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    learning_rate=2e-4,
    num_train_epochs=2,
    #evaluation_strategy="epoch",
    logging_steps=100,
    logging_dir="./logs",
    #save_strategy="epoch",
    #save_total_limit=2,
    fp16=False,
    bf16=True,
    push_to_hub=False,
    report_to="none"
)

# 10. Trainer Setup
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_data["train"],
    eval_dataset=tokenized_data["test"],
    tokenizer=tokenizer,
    data_collator=data_collator
)

```

```python
# 11. Train
trainer.train()
```

```python
# 12. Save the Model
model.save_pretrained("./tinyllama-lora-sft-t1")
tokenizer.save_pretrained("./tinyllama-lora-sft-t1")
```
